````python
# 🍾 Python Loops Demo - Old Monk Example

bottle_ml = 750
friends = ["Manoj", "Rahul", "Ravi", "Kiran"]
peg_size = 30  # 30ml per peg
rounds = 7

print(f"🍾 Bottle Size : {bottle_ml} ml")
print(f"👥 Friends     : {len(friends)}")
print(f"🥃 Peg Size    : {peg_size} ml")
print("-" * 50)


for round_no in range(1,rounds+1):
    print(f"🔁 Round {round_no}")

    for friend in friends:

        if bottle_ml < peg_size:
            print("❌ Bottel is empty eddu manage hogiii....")
            print("Party is over ....")
            break

       # bottle_ml = bottle_ml - peg_size
        bottle_ml -= peg_size

        print(f"{friend} takes one 🥃 | and Remaining = {bottle_ml}ml")

    if bottle_ml < peg_size:
        break


print("-"*50)
print(f"Bottle Remaining : {bottle_ml} ml")
print("Demo Completed ✅")
    
````

````python
# Student database — a dictionary of dictionaries!
students_db = {
    "STU001": {
        "name"  : "Rahul Gupta",
        "marks" : {"Maths": 88, "Science": 92, "English": 79},
        "grade" : "A",
    },
    "STU002": {
        "name"  : "Sneha Patel",
        "marks" : {"Maths": 95, "Science": 97, "English": 91},
        "grade" : "A+",
    },
    "STU003": {
        "name"  : "Arjun Singh",
        "marks" : {"Maths": 62, "Science": 70, "English": 68},
        "grade" : "B",
    },
}
````

````python
from numpy import average


print("\n📊 STUDENT REPORT CARD")
print("=" * 50)

for student_id,info in students_db.items():
    name = info['name']
    marks = info['marks']
    grade = info['grade']
    
    total = sum(marks.values())
    average = total / len(marks)

    print(f"\n  {student_id} — {name}  [{grade}]")
    for subject, score in marks.items():
        bar = "█" * (score // 10)
        print(f"    {subject:<10}: {score:3d}  {bar}")
    print(f"    {'Average':<10}: {average:.1f}")
````
