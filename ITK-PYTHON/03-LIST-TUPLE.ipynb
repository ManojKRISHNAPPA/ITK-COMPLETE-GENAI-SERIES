```python
# =============================================
# 🛠️  USEFUL LIST OPERATIONS
# =============================================

ipl_scores = [165, 182, 143, 201, 155, 178, 190, 134]

# Statistics
print(f"Scores     : {ipl_scores}")
print(f"Count      : {len(ipl_scores)}")
print(f"Highest    : {max(ipl_scores)}")
print(f"Lowest     : {min(ipl_scores)}")
print(f"Total runs : {sum(ipl_scores)}")
print(f"Average    : {sum(ipl_scores)/len(ipl_scores):.1f}")

# Sorting (creates NEW list, original unchanged)
sorted_scores = sorted(ipl_scores)               # ascending
top3          = sorted(ipl_scores, reverse=True)[:3]  # top 3

print(f"\nSorted     : {sorted_scores}")
print(f"Top 3      : {top3}")

# Check if above 150
high_scores = [s for s in ipl_scores if s > 150]    # list comprehension — preview!
print(f"Above 150  : {high_scores}")

# Copy a list (IMPORTANT — don't just use = )
original  = [1, 2, 3]
wrong_copy = original         # This is NOT a copy — it's the SAME list!
right_copy = original.copy()  # This is a real copy

wrong_copy.append(4)
print(f"Original   : {original}")    # [1, 2, 3, 4] — CHANGED! Bug!

original = [1, 2, 3]
right_copy = original.copy()
right_copy.append(4)
print(f"Original   : {original}")    # [1, 2, 3] — unchanged ✅
print(f"Right copy : {right_copy}")  # [1, 2, 3, 4] ✅
```

```python
# =============================================
# 🍕 FULL ZOMATO ORDER SIMULATOR
# Lists + Tuples + Input + Math all together!
# =============================================

print("=" * 45)
print("🍕 SWIGGY ORDER SIMULATOR...")
print("=" * 45)

MENU = [
    ("BUTTER CICKEN",   280),
    ("PANNER TIKKA",    220),
    ("Garlic Naan",      40),
    ("Dal Makhani",     180),
    ("Biryani",         320),
    ("Lassi",            60),
    ("Gulab Jamun",      80),
    ("Raita",            50),
]



# show menu to customer:
print("\n 📋 MENU")
for i, (item,price) in enumerate(MENU,start=1):
    print(f" {i}. {item:<20} Rs {price}")

cart = []

print("\n ➕ ADD ITEMS TO CART (enter number, or 0 to finsh):")


while True:
    choice =  input("Item number : ")

    if not choice.isdigit():
        print("please enter valid number..")
        continue

    choice = int(choice)

    if choice == 0:
        break

    if 1<= choice <=len(MENU):
        selected = MENU[choice - 1]
        cart.append(selected)
        print(f"✅ added: {selected[0]} - Rs {selected[1]}")
    else: 
        print(f"  ❌ Invalid choice. Enter 1 to {len(MENU)}")


# Show order summary
if not cart:
    print("\n❌ No items in cart. Goodbye!")
else:
    subtotal   = sum(price for item, price in cart)
    gst        = int(subtotal * 0.05)    # 5% food GST
    delivery   = 40
    grand_total = subtotal + gst + delivery

    print("\n" + "=" * 45)
    print("          📦 ORDER SUMMARY")
    print("=" * 45)

    for i, (item, price) in enumerate(cart, start=1):
        print(f"  {i}. {item:<22} Rs {price}")

    print("-" * 45)
    print(f"  {'Subtotal':<25} Rs {subtotal}")
    print(f"  {'GST (5%)':<25} Rs {gst}")
    print(f"  {'Delivery Charge':<25} Rs {delivery}")
    print("=" * 45)
    print(f"  {'GRAND TOTAL':<25} Rs {grand_total}")
    print("=" * 45)
    print(f"\n  🛵 Estimated delivery: 30-45 minutes")
    print(f"  📱 Order placed successfully!")


```
