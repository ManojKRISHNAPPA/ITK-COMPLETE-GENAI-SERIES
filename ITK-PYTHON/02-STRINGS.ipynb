```python
# =============================================
# 🎨 STRING FORMATTING — MAKING BEAUTIFUL OUTPUT
# =============================================

# ---- f-string formatting codes ----
runs    = 12040
avg     = 58.0672
name    = "Virat"

# Basic
print(f"Runs: {runs}")                    # Runs: 12040
```

```python
# Width alignment
print(f"{'Name':<15} {'Runs':>8} {'Avg':>8}")   # Headers
print(f"{'Virat Kohli':<15} {12040:>8} {58.07:>8.2f}")
print(f"{'Rohit Sharma':<15} {9500:>8} {48.61:>8.2f}")
print(f"{'MS Dhoni':<15} {10773:>8} {50.57:>8.2f}")
```

```python
# Number formatting
price = 1234567.89
print(f"Price  : Rs {price:,.2f}")           # Rs 1,234,567.89 (comma separator)
print(f"Score  : {0.956:.1%}")               # 95.6% (percentage)
print(f"Binary : {255:b}")                   # 11111111
print(f"Padded : {7:03d}")  
                 # 007 (zero-padded)
```

```python
# WHATSAPP MESSAGE FORMATER

print("="*45)
print(" 📲 WHATSAPP MESSAGE FORMATTER ")
print("="*45)

#TAKE INPUT
sender = input("Enter your name: ").strip().title() #Clean + Capitilaisation
reciver = input("Friends name: ").strip().title()
message = input("Your message: ").strip()


# analysis
word_count = len(message.split())
char_count = len(message)
char_with_nospace = len(message.replace(' ',''))

# Check if it's too formal (contains sir/ma'am)
is_formal = any(word in message.lower() for word in ["sir", "ma'am", "madam", "dear"])

# Check if it has emojis (simplified check)
has_emojis = any(char for char in message if ord(char) > 127)


# Format the message
first_word  = message.split()[0] if message else ""
preview     = message[:30] + "..." if len(message) > 30 else message

print()
print("─" * 45)
print(f"  From    : {sender}")
print(f"  To      : {reciver}")
print(f"  Preview : {preview}")
print(f"  Words   : {word_count}")
print(f"  Chars   : {char_count} (without spaces: {char_with_nospace})")
print(f"  Formal  : {'Yes' if is_formal else 'No'}")
print(f"  Emojis  : {'Yes 😄' if has_emojis else 'No'}")
print("─" * 45)

# Generate WhatsApp-style formatted output
print(f"\n  💬 {sender} → {reciver}")
print(f"  {message}")
print(f"  ✓✓ 10:47 AM")
print()

# Suggest improvements
if word_count < 3:
    print("  💡 Tip: Message is very short. Add more detail!")
if message == message.upper() and len(message) > 5:
    print("  💡 Tip: Writing in ALL CAPS feels like shouting 😅")
if not message.endswith((".", "!", "?", "😊", "😄")):
    print("  💡 Tip: Message has no ending punctuation.")
```
