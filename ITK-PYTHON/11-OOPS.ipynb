# Inheritenace

### Single Inheritence

```python
# parent class:
class Car:
    def __init__(self,windows,doors,engine) -> None:
        self.windows = windows
        self.doors = doors
        self.engine = engine
    
    def drive(self):
        print(f"the person will drive {self.engine} car")
```

```python
car1 = Car(4,5,"Petrol")
```

```python
car1.drive()
```

```python
# Child class
class TataEv(Car):
    def __init__(self, windows, doors, engine,is_selfdriving):
        super().__init__(windows, doors, engine)
        self.is_selfdriving = is_selfdriving       

    def check_is_selfdriving(self):
        print(f"TataEV supports self driving: {self.is_selfdriving}")
```

```python
tata = TataEv(4,5,"electric",True)
```

```python
tata.check_is_selfdriving()
```

```python
tata.drive()
```

## Key Points to Remember
1. Inheritance lets one class reuse another class's code.

2. Parent Class = Base Class = Super Class (same meaning).

3. Child Class = Derived Class = Sub Class (same meaning).

4. super() calls the parent class's method, most commonly its constructor.

5. It avoids duplicate code and follows the DRY (Don't Repeat Yourself) principle.

6. A child class can add new features or override existing ones.

7. Single Inheritance means one parent class; Multiple Inheritance means more than one parent class.

# Polymorphism

Polymorphism is a core concept in Object-Oriented Programming (OOP) that allows objects of different classes to be treated as objects of a common superclass. It provides a way to perform a single action in different forms. Polymorphism is typically achieved through method overriding and interfaces

```python
# Parent class:

class BankAccount:

    def __init__(self,account_holder,balance):
        self.account_holder = account_holder
        self.balance = balance

    def account_details(self):
        print("--------Acccount Details---------")
        print(f"Account holder name: {self.account_holder} ")
        print(f"Balance: {self.balance}")

    def calculate_intrest(self):
        print("Default intrest calculation from Bankaccount")
```

```python
class SavingAccount(BankAccount):
    def calculate_intrest(self):
        intrest = self.balance * 0.04

        print("saving account intrest")
        print(f"Intrest = Rs {intrest}")

    
```

```python
class CurrentAccount(BankAccount):
    def calculate_intrest(self):
        print("current account intrest no intrest has been calculated")
```

```python
class FixedDeposit(BankAccount):
    def calculate_intrest(self):
        intrest = self.balance * 0.075

        print("fixed deposit account intrest")
        print(f"Intrest = Rs {intrest}")
```

```python
saving = SavingAccount("Manoj",20000)
```

```python
saving.calculate_intrest()
```

```python
current = CurrentAccount("Arun",300000)
```

```python
current.calculate_intrest()
```

```python
fixed  = FixedDeposit("pramod",100000)
```

```python
fixed.account_details()
```

```python
fixed.calculate_intrest()
```

# Encapsulation

```python
# wrapping data and methods together into one unit and protecting data from direct modification

class BankAccount:
    def __init__(self,owner,balance) -> None:
        self.owner = owner
        self.balance = balance

    def deposit(self,ammount):
        self.balance += ammount

    def show_balance(self):
        print(f"Balance =  Rupees{self.balance}")
        
```

```python
account = BankAccount("Manoj",10000)
```

```python
account.show_balance()
```

```python
account.deposit(2000)
```

```python
account.show_balance()
```

```python
account.balance = -150000
```

```python
account.show_balance()
```

```python

class Wallet:

    def __init__(self):
        self.__balance = 0
    
    def add_money(self,amount):

        if amount > 0:
            self.__balance += amount
        
    def spend(self,amount):
        if amount<=self.__balance:
            self.__balance -= amount
        else:
            print("Inssuficient funds") 

    def get_balance(self):
        return self.__balance
```

```python
# Property Decorator

# @property
```

```python
class Employee:

    def __init__(self,salary):

        self._salary=salary

    @property
    def salary(self):
        return self._salary
    

    @salary.setter
    def salary(self,value):
        if value > 0:
            self._salary = value
```

```python
emp = Employee(50000)
```

```python
print(emp.salary)
```

```python
emp.salary = 7000
```

```python
from abc import ABC, abstractmethod

class Vehical(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

```

```python
class Car(Vehical):
    def start(self):
        print("Car started")
    
    def stop(self):
        print("Car Stopeed")
```

```python
Vehical()
```

```python
class Payment(ABC):
    
    @abstractmethod
    def pay(self,amount):
        pass
```

```python
class UPI(Payment):

    def pay(self,amount):
        print(f"Paid ₹{amount} using UPI")
```

```python
class CreditCard(Payment):

    def pay(self,amount):
        print(f"Paid ₹{amount} using Card")
```

```python
payament = UPI()
```

```python
payament.pay(120)
```
