# Inheritance In Python

# SINGLE Inheritance

````python
# PARENT CLASS:
class Car:
    def __init__(self,windows,doors,engine):
        self.windows=windows
        self.engine=engine
        self.doors=doors

    def drive(self):
        print(f"the person will drive {self.engine} car")
````

````python
car1 = Car(4,5,"Petrol")
````

````python
car1.drive()
````

````python
class TATA_EV(Car):
    def __init__(self, windows, doors, engine,is_selfdrive):
        super().__init__(windows, doors, engine)
        self.is_selfdrive = is_selfdrive

    def is_selfdrive(self):
        return f"TATA EV supports self driving : {self.is_selfdrive}"
````

````python
tata = TATA_EV(4,5,"electtric",True)
````

````python
tata.drive()
````

````python
tata.is_selfdrive
````

# Key Points to Remember
    1. Inheritance lets one class reuse another class's code.

    2. Parent Class = Base Class = Super Class (same meaning).

    3. Child Class = Derived Class = Sub Class (same meaning).

    4. super() calls the parent class's method, most commonly its constructor.

    5. It avoids duplicate code and follows the DRY (Don't Repeat Yourself) principle.

    6. A child class can add new features or override existing ones.
    
    7. Single Inheritance means one parent class; Multiple Inheritance means more than one parent class.

# Polymorphism
Polymorphism is a core concept in Object-Oriented Programming (OOP) that allows objects of different classes to be treated as objects of a common superclass. It provides a way to perform a single action in different forms. Polymorphism is typically achieved through method overriding and interfaces

# What is Method Overriding?

Imagine a bank.

Every account can calculate interest.

But...

Savings Account → 4% Interest

Current Account → No Interest

Fixed Deposit → 7.5% Interest

All of them have the same method:

```
calculate_interest()
```

````python
# Parent class
class BankAccount:

    def __init__(self,account_holder,balance):
        self.account_holder = account_holder
        self.balance = balance

    def account_details(self):
        print("------ Account Details ------")
        print(f"Account Holder : {self.account_holder}")
        print(f"Balance        : ₹{self.balance}")
    
    def calculate_interest(self):
        print("Default interest calculation from BankAccount") 

# savings account
class SavingsAccount(BankAccount):
    def calculate_interest(self):
        intrest = self.balance * 0.04

        print("savings account intrest")
        print(f"Intrest = Rs {intrest}")

class CurrentAccount(BankAccount):
    def calculate_interest(self):

        print("current savings account")
        print(f"no intrest is caluclated")

class FixedDeposit(BankAccount):
    def calculate_interest(self):
        intrest = self.balance * 0.075

        print("fixed deposit account intrest")
        print(f"Intrest = Rs {intrest}")    
````

````python
savings = SavingsAccount("Manoj",100000)
current = CurrentAccount("Arun",200000)
fd = FixedDeposit("pradeep",21000)
````

````python
savings.account_details()
````

````python
current.account_details()
````

````python
fd.account_details()
````

````python
savings.calculate_interest()
````

````python
# Getter Method:

class BankAccount:

    def __init__(self, name, balance):
        self.__name = name
        self.__balance = balance
    
    def get_balance(self):
        return self.__balance

# setter methood:
    
    def set_balance(self,balance):
        if balance >= 0:
            self.__balance = balance
        
        else:
            print("Invalid balance")
        
````

````python
account = BankAccount("manoj",10000)

account.set_balance(5000)

print(account.get_balance())
````

````python
class BankAccount:

    def __init__(self,balance):
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance
````

````python
class BankAccount:

    def __init__(self,balance):
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance
    
    @balance.setter
    def balance(self,value):

        if value >= 0:
            self.__balance = value
        else:
            print("Invalid Balance")
````

````python
account = BankAccount(10000)

account.balance = 5000
````

````python
# without property
account.get_balance()


# with proeperty

account.balance = 5000
````

````python
from abc import ABC, abstractmethod
````

````python
# ABC --> abstarct base class
````

````python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass
````

````python
class UPI(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")
````

````python
class CreditCard(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using Card")
````

````python
upi = UPI()
````

````python
upi.pay(1000)
````

````python
payment = Payment()
````
