```python
#  PizzA PRICE CALCULATION

def calculate_pizza_price(size):
    """ calculate the price of a pizza based on its size """

    base_price = 0

    if size == "small":
        base_price = 199
    elif size == "medium":
        base_price = 299
    elif size == "large":
        base_price = 399
    else:
        print("Invalid size")
        return None

    total_price = base_price
    return total_price
```

```python
price1 = calculate_pizza_price("small")
price2 = calculate_pizza_price("medium")
price3 = calculate_pizza_price("large")
```

```python
print(f"Price of small pizza: {price1}")
print(f"Price of medium pizza: {price2}")       
print(f"Price of large pizza: {price3}")
```
