````python
#  PizzA PRICE CALCULATION

def calculate_pizza_price(size):
    """ calculate the price of a pizza based on its size """

    base_price = 0

    if size == "small":
        base_price = 199
    elif size == "medium":
        base_price = 299
    elif size == "large":
        base_price = 399
    else:
        print("Invalid size")
        return None

    total_price = base_price
    return total_price
````

# The Full Pizza Ordering System

````python
# =============================================
# 🍕 PIZZA ORDERING SYSTEM — ALL FUNCTIONS TOGETHER
# =============================================

# Menu constants
SIZES     = {"S": ("Small",  199), "M": ("Medium", 299), "L": ("Large",  399)}
TOPPINGS  = {"1": ("Pepperoni", 60), "2": ("Paneer",   50),
             "3": ("BBQ Chicken", 70), "4": ("Corn & Capsicum", 40)}
OFFERS    = {"WELCOME10": 10, "SAVE20": 20, "STUDENT15": 15}


def show_menu():
    print("\n📋 PIZZA MENU")
    print("─" * 40)
    print("SIZES:")
    for code, (name, price) in SIZES.items():
        print(f"  [{code}] {name:<10} Rs {price}")
    print("\nTOPPINGS:")
    for code, (name, price) in TOPPINGS.items():
        print(f"  [{code}] {name:<20} +Rs {price}")


def get_size():
    while True:
        choice = input("Select size (S/M/L): ").upper()
        if choice in SIZES:
            return SIZES[choice]
        print("  ❌ Invalid. Enter S, M, or L")

def get_toppings():
    selected = []
    print("Select toppings (enter numbers, comma separated, or press Enter to skip):")
    choices = input("Toppings: ").split(",")
    for c in choices:
        c = c.strip()
        if c in TOPPINGS:
            selected.append(TOPPINGS[c])
    return selected


def get_quantity():
    while True:
        qty = input("Quantity: ")
        if qty.isdigit() and int(qty) > 0:
            return int(qty)
        print("  ❌ Enter a valid number")

def apply_coupon(subtotal):
    code = input("Coupon code (or press Enter to skip): ").upper()
    if code in OFFERS:
        disc_pct = OFFERS[code]
        discount = subtotal * disc_pct / 100
        print(f"  🎉 Coupon applied! {disc_pct}% off — saved Rs {discount:.0f}")
        return discount
    elif code:
        print("  ❌ Invalid coupon code.")
    return 0

def print_bill(order_details):
    print("\n" + "=" * 45)
    print("          🍕 DOMINO'S PYTHON — RECEIPT")
    print("=" * 45)
    for item in order_details["items"]:
        print(f"  {item['desc']:<32} Rs {item['price']}")
    print("─" * 45)
    print(f"  {'Subtotal':<32} Rs {order_details['subtotal']:.0f}")
    if order_details["discount"] > 0:
        print(f"  {'Discount':<32} -Rs {order_details['discount']:.0f}")
    print(f"  {'GST (5%)':<32} Rs {order_details['gst']:.0f}")
    print(f"  {'Delivery Charge':<32} Rs 40")
    print("=" * 45)
    print(f"  {'GRAND TOTAL':<32} Rs {order_details['total']:.0f}")
    print("=" * 45)
    print(f"\n  🛵 Estimated time: 30-40 minutes")
    print(f"  📱 Order #{order_details['order_id']} placed successfully!\n")

def take_order(order_id):
    print(f"\n🍕 NEW ORDER — #{order_id}")
    show_menu()

    size_name, size_price = get_size()
    toppings              = get_toppings()
    quantity              = get_quantity()

    # Build items list
    items  = []
    unit   = size_price + sum(p for _, p in toppings)
    topping_names = ", ".join(n for n, _ in toppings) if toppings else "No topping"

    items.append({
        "desc":  f"{size_name} Pizza ({topping_names}) x{quantity}",
        "price": unit * quantity
    })

    subtotal = sum(i["price"] for i in items)
    discount = apply_coupon(subtotal)
    after_discount = subtotal - discount
    gst   = after_discount * 0.05
    total = after_discount + gst + 40

    return {
        "order_id": order_id,
        "items":    items,
        "subtotal": subtotal,
        "discount": discount,
        "gst":      gst,
        "total":    total,
    }

````

````python
## Main Program

print("🍕 welcome to ITKs Dominos pizza")

orders = []

while True:
    print("\n1. Place Order")
    print("2. View order history")
    print("3. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        order = take_order(len(orders) + 1001)
        print_bill(order)
        orders.append(order)

    elif choice == "2":
        if not orders:
            print("No orders yet.")
        else:
            print(f"\n📋 {len(orders)} orders today:")
            for o in orders:
                print(f"  #{o['order_id']} — Rs {o['total']:.0f}")
            print(f"  Total revenue: Rs {sum(o['total'] for o in orders):.0f}")
    elif choice == "3":
        print("Thank you for using ITKs Dominos pizza. Goodbye!")
        break
````
