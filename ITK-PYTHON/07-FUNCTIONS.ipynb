```python
# without function

gst1 = 450 * 0.18
total1 = 450 + gst1

print(total1)

gst2 = 200 * 0.18
total2 = 200 + gst2

print(total2)

```

```python
#with fucntion 

def calculuate_total(amount):
    gst = amount * 0.18

    return amount + gst
```

```python
calculuate_total(400)
```

```python
calculuate_total(200)
```
