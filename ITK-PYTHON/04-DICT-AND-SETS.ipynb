````python
student_db = {
    "std-1":{
        "name" : "Manoj",
        "marks" : {"maths": 88, "Science": 87,"English": 60},
        "Grade" : "A",
    },

    "std-2":{
        "name" : "Meghana",
        "marks" : {"maths": 88, "Science": 87,"English": 60},
        "Grade" : "B",
    },
    "std-3":{
        "name" : "Arun",
        "marks" : {"maths": 88, "Science": 87,"English": 80},
        "Grade" : "A",
    },
    "std-4":{
        "name" : "Prajna",
        "marks" : {"maths": 88, "Science": 87,"English": 60},
        "Grade" : "C",
    }
}
````

````python
student_db["std-1"]
````

````python
student_db["std-1"]["name"]
````

````python
student_db["std-1"]["marks"]
````

````python
student_db["std-1"]["marks"]["maths"]
````

````python
for student_id,info in student_db.items():
    name = info["name"]
    marks =  info["marks"]
    Grade = info["Grade"]
    total =  sum(marks.values())
    avg = total / len(marks)

    print(f" {student_id} and -{name} [{Grade}]")
    for subject, score in marks.items():
        bar = "█" * (score // 10)
        print(f"    {subject:<10}: {score:3d}  {bar}")
    print(f"    {'Average':<10}: {avg:.1f}")
````

````python
# =============================================
# 🍽️ SIMPLE RESTAURANT MANAGEMENT SYSTEM
# Lists + Dictionaries + Sets
# =============================================

# Menu dictionary

menu = {
    "Paneer Butter Masala": 220,
    "Veg Biryani": 200,
    "Butter Chicken": 280,
    "Chicken Biryani": 300,
    "Garlic Naan": 40,
    "Lassi": 60,
    "Cold Coffee": 80
}

order = [] #  list it stores the orderd items

# set 
customers = set() # stores the unique customers

# Dictionary

sales = {}

print("🍽️ Welcome to ITK RESTARUANT")

customer = input("enter the customer name: ").title()
customers.add(customer)


print("----------MENU_______")
for item,price in menu.items():
    print(f"{item:<25} price is RS{price}")

print("Type 'done' to finish order" )

while True:
    item = input("Enter item: ").title()

    if item.lower() == "done":
        break

    if item in menu:
        order.append(item)

        # count sales
        if item in sales:
            sales[item] += 1
        else:
            sales[item] = 1

        print("all the items added ✅")

#----- BILL-----
total = 0

print("\n========== BILL ==========")

for item in order:
    print(f"{item:<25} Rs {menu[item]}")
    total += menu[item]

print("--------------------------")
print("Total Bill :", total)

# ------------- SUMMARY ----------------

print("\n========== SUMMARY ==========")

print("Customer :", customer)
print("Items Ordered :", order)
print("Unique Customers :", customers)
print("Sales Dictionary :", sales)

if sales:
    best_item = max(sales, key=sales.get)
    print("Best Selling Item :", best_item)

print("Thank You 😊")
````

Concept Summary

| Structure | Syntax | Use When |
|-----------|--------|---------|
| List `[]` | `[1, 2, 3]` | Ordered items, can change |
| Tuple `()` | `(1, 2, 3)` | Ordered items, cannot change |
| Dictionary `{}` | `{"key": value}` | Label → value lookup |
| Set `{}` | `{1, 2, 3}` | Unique items only |

**Dictionary Methods:**

| Method | Does |
|--------|------|
| `d["key"]` | Get value (crashes if missing) |
| `d.get("key", default)` | Safe get (returns default) |
| `d["key"] = val` | Set / update value |
| `del d["key"]` | Delete key |
| `d.pop("key")` | Delete and return value |
| `d.keys()` | All keys |
| `d.values()` | All values |
| `d.items()` | All key-value pairs |
| `d.update(d2)` | Merge d2 into d |
| `"key" in d` | Check if key exists |
