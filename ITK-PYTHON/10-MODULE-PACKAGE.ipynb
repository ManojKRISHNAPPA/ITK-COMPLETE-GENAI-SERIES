```python
# calculate avg using python code

from numpy import average


marks = [85,72,90,60,70]
total = 0

for mark in marks:
    total = total + mark

average = total / len(marks)

print(f"Average:  {average}")
```

```python
import statistics

marks = [85,72,90,60,70]

average = statistics.mean(marks)

print(f"Average:  {average}")
```
