# Optional

```python
user = {
    1: "Manoj",
    2: "Kumar",
    3: "ITkannadigaru"
}
```

```python
def find_user(user_id: int) -> str:
    return user.get(user_id)
```

```python
find_user(1)
```

```python
find_user(100)
```

```python
from typing import Optional

def find_user(user_id: int) -> Optional[str]:
    return user.get(user_id)
```

```python
from typing import Optional

def find_user(user_id: int) -> str | None:
    return user.get(user_id)
```

```python
find_user(1000)
```

```python
from typing import Union

def parse_id(value: Union[int, str]) -> int:
    return int(value)
```

```python

def parse_id(value: str | int) -> int:
    return int(value)
```

```python
from typing import Literal

LogLevel = Literal["DEBUG", "INFO", "WARNING", "ERROR", "CRITICAL"]
```

```python
def log(message:str, level: LogLevel) -> None:
    print(f"[{level}] {message}")
```

```python
log("This is a debug message.", "DEBUG")
```

```python
from typing import TypedDict

class ServerConfig(TypedDict):
    host: str
    port: int
    use_ssl: bool

def connect_to_server(config: ServerConfig) -> None:
    print(f"Connecting to {config['host']} on port {config['port']} with SSL: {config['use_ssl']}")
```
