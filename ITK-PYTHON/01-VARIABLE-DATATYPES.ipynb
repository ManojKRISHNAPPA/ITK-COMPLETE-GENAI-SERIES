````python
# =============================================
# ⌨️  USER INPUT — MAKING PROGRAMS INTERACTIVE
# =============================================

# input() always returns a STRING
# You must convert it if you need math

print("🏏 CRICKET BATTING AVERAGE CALCULATOR")
print("=" * 40)

player_name   = input("Enter player name: ")
total_runs    = int(input("Enter total runs: "))
total_innings = int(input("Enter total innings: "))
not_outs      = int(input("Enter not outs: "))

# Calculate
actual_innings = total_innings - not_outs
batting_avg    = total_runs / actual_innings

print()
print(f"Player   : {player_name}")
print(f"Total Runs : {total_runs}")
print(f"Innings  : {actual_innings}")
print(f"Average  : {batting_avg:.2f}")

# Grade the average
if batting_avg >= 50:
    grade = "World Class 🌟"
elif batting_avg >= 40:
    grade = "Very Good ✅"
elif batting_avg >= 30:
    grade = "Average 👍"
else:
    grade = "Needs Work 💪"

print(f"Rating   : {grade}")
````

# Concept Summary Table

| Concept | What it is | Example |
|---------|-----------|---------|
| Variable | A labelled box to store data | `score = 100` |
| `str` | Text in quotes | `"Virat Kohli"` |
| `int` | Whole number | `35` |
| `float` | Decimal number | `58.07` |
| `bool` | True or False | `True` |
| `None` | Empty / not set | `None` |
| `type()` | Find what type a variable is | `type(score)` |
| `int()` | Convert to integer | `int("42")` |
| `float()` | Convert to decimal | `float("3.14")` |
| `str()` | Convert to text | `str(100)` |
| `input()` | Get text from user | `input("Name: ")` |
| f-string | Smart print with variables | `f"Hello {name}"` |
