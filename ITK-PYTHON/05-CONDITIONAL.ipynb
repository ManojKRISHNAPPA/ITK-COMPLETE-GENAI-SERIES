````python
# ---- Complex ATM condition ----
balance      = 10000
daily_limit  = 20000
withdrawn_today = 5000
amount       = 8000

# Can withdraw if: amount available AND within daily limit AND amount > 0
can_withdraw = (amount > 0 and
                amount <= balance and
                (withdrawn_today + amount) <= daily_limit)

if can_withdraw:
    print("✅ Transaction allowed")
else:
    print("❌ Transaction denied")
    if amount > balance:
        print("   Reason: Insufficient balance")
    if (withdrawn_today + amount) > daily_limit:
        print("   Reason: Daily limit exceeded")
````

````python
# =============================================
# 🏧 ATM MACHINE — FULL SIMULATION
# (Nested conditions)
# =============================================

print("=" * 40)
print("    🏧 WELCOME TO  ITK PYTHON BANK ATM")
print("=" * 40)

# Account info
accounts = {
    "123456": {"pin": "1234", "balance": 15000, "name": "Rahul"},
    "789012": {"pin": "5678", "balance": 8500,  "name": "Priya"},
    "789013": {"pin": "1289", "balance": 20000,  "name": "ITK"},
}

card_number = input("\nEnter Card Number: ").strip()

if card_number in accounts:
    account = accounts[card_number]

    # PIN check (max 3 attempts)
    attempts = 0
    pin_correct = False

    while attempts < 3:
        pin = input("Enter PIN: ").strip()
        if pin == account["pin"]:
            pin_correct = True
            break
        else:
            attempts += 1
            remaining = 3 - attempts
            if remaining > 0:
                print(f"❌ Wrong PIN. {remaining} attempt(s) remaining.")
            else:
                print("🚫 Card BLOCKED! Too many wrong attempts.")

    if pin_correct:
        print(f"\n✅ Welcome, {account['name']}!")

        # Main menu
        while True:
            print("\n1. Check Balance")
            print("2. Withdraw Cash")
            print("3. Deposit Money")
            print("4. Exit")

            choice = input("\nSelect option: ").strip()

            if choice == "1":
                print(f"\n💰 Available Balance: Rs {account['balance']:,}")

            elif choice == "2":
                amount_str = input("Enter amount (multiple of 100): Rs ")

                if not amount_str.isdigit():
                    print("❌ Invalid input. Numbers only.")
                else:
                    amount = int(amount_str)

                    if amount <= 0:
                        print("❌ Amount must be positive.")
                    elif amount % 100 != 0:
                        print("❌ Must be in multiples of 100.")
                    elif amount > 10000:
                        print("❌ Max Rs 10,000 per transaction.")
                    elif amount > account["balance"]:
                        print(f"❌ Insufficient funds. Balance: Rs {account['balance']:,}")
                    else:
                        account["balance"] -= amount
                        print(f"✅ Rs {amount:,} dispensed. Remaining: Rs {account['balance']:,}")

            elif choice == "3":
                deposit = input("Enter deposit amount: Rs ")
                if deposit.isdigit() and int(deposit) > 0:
                    account["balance"] += int(deposit)
                    print(f"✅ Rs {int(deposit):,} deposited. New balance: Rs {account['balance']:,}")
                else:
                    print("❌ Invalid amount.")

            elif choice == "4":
                print("\n👋 Thank you for using Python Bank ATM!")
                print("   Please collect your card. Have a great day!")
                break

            else:
                print("❌ Invalid option. Choose 1-4.")

else:
    print("❌ Card not recognized. Please try another card.")
````

## 🎯 Module 05 — Conditions Cheat Sheet

| Operator | Meaning | Example |
|----------|---------|---------|
| `==` | Equal | `score == 100` |
| `!=` | Not equal | `status != "blocked"` |
| `>` | Greater than | `balance > 0` |
| `<` | Less than | `attempts < 3` |
| `>=` | Greater or equal | `age >= 18` |
| `<=` | Less or equal | `amount <= limit` |
| `and` | Both must be True | `age >= 18 and has_id` |
| `or` | Either must be True | `is_vip or paid_full` |
| `not` | Flip True/False | `not is_blocked` |
| `in` | Exists in collection | `city in allowed_cities` |
| `not in` | Doesn't exist | `item not in cart` |
| `is` | Same object | `x is None` |
| `is not` | Not same object | `x is not None` |
