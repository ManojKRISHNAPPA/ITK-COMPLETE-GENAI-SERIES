```python
# TUPLE 

categories = (
    "python",
    "devops",
    "cloud",
    "ai",
    "database"
)


#list

books = [
    "Python Basics",
    "Docker Deep Dive",
    "Kubernetes",
    "Terraform",
    "AWS",
    "Python Basics",
    "GenAI"
]


# set

members = {
    "Rahul",
    "Amit",
    "Sneha",
    "Rahul",
    "Priya"
}

#DICTIONARY

library = {
    "Python Basics": 5,
    "Docker Deep Dive": 2,
    "Kubernetes": 3,
    "Terraform": 1,
    "AWS": 4,
    "GenAI": 2
}

while True:

    print("\n")
    print("=" * 60)
    print("📚 SMART LIBRARY MANAGEMENT SYSTEM")
    print("=" * 60)

    print("1. Show Categories")
    print("2. Show All Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Show Available Books")
    print("7. Show Members")
    print("8. Show Unique Books")
    print("9. Count Each Book")
    print("10. Low Stock Alert")
    print("11. Search Books by Starting Letter")
    print("12. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("Categories:")
        for category in categories:
            print(f"- {category}")

    elif choice == "2":
        print("All Books:")
        for book in books:
            print(f"- {book}")

    elif choice == "3":
        search_term = input("Enter a book name to search for: ")
        found = False
        for book in books:
            if search_term.lower() in book.lower():
                print(f"Found: {book}")
                found = True
        if not found:
            print("Book not found")

    elif choice == "4":
        borrow_book = input("Enter the book name to borrow: ")
        if borrow_book in library:
            library[borrow_book] -= 1
            print(f"Borrowed: {borrow_book}")
        else:
            print("Book not available")

    elif choice == "5":
        return_book = input("Enter the book name to return: ")
        if return_book in library:
            library[return_book] += 1
            print(f"Returned: {return_book}")   

    elif choice == "6":
        print("Available Books:")
        for book, count in library.items():
            if count > 0:
                print(f"{book} - {count} copies available")
    elif choice == "7":
        print("Members:")
        for member in members:
            print(f"- {member}")

    elif choice == "8":
        print("Unique Books:")
        for book in set(books):
            print(f"- {book}")
    elif choice == "9":
        print("Count of Each Book:")
        for book, count in library.items():
            print(f"{book}: {count}")

    elif choice == "10":
        print("Low Stock Alert:")
        for book, count in library.items():
            if count < 3:
                print(f"{book} - {count} copies available")
    elif choice == "11":
        start_letter = input("Enter a starting letter: ")
        print(f"Books starting with '{start_letter}':")
        for book in books:
            if book.startswith(start_letter):
                print(f"- {book}")  
    elif choice == "12":
        print("Exiting the program...")
        break
    else:
        print("Invalid choice. Please try again.")







```
